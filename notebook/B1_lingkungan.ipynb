```python
# Kode 1.1 Uji Lingkungan dan Struktur Larik Data

import sys
sys.path.append("src")

import cv2
import numpy as np
import skimage
import matplotlib.pyplot as plt
from utils_cv import baca, tampil

print("OpenCV:", cv2.__version__, "| NumPy:", np.__version__, "| scikit-image:", skimage.__version__)

img = baca("data/sintetis/citra/teh_001.png")
print("shape:", img.shape, "dtype:", img.dtype)
print("min-maks:", img.min(), img.max())
print("piksel (y=100, x=200) dalam urutan BGR:", img[100, 200])

plt.figure(figsize=(10, 4))
plt.subplot(1, 2, 1)
plt.imshow(img)
plt.title("Tanpa konversi (salah)")

plt.subplot(1, 2, 2)
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title("Dengan konversi BGR ke RGB")

plt.savefig("hasil/b1_bgr_rgb.png")
plt.show()
```
